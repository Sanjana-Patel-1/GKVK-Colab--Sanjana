```python
import pandas as pd
data={
    'department':['HR','sales','IT','sales','IT','HR','sales'],
    'employee':['A','B','C','D','E','F','G'],
    'salary':['50000','60000','45000','48000','70000','72000','71000'],
    'experience':['2','4','3','5','6','7','5'],

}
df=pd.DataFrame(data)

df['salary'] = pd.to_numeric(df['salary'])
df['experience'] = pd.to_numeric(df['experience'])

print(df)

df1=df.groupby('experience').sum()
print(df1)

df2=df.groupby(['department','experience']).sum()
print(df2)

df3=df.groupby('department')['salary'].agg(['sum','mean'])
print(df3)

df4=df['dept_avg_salary']=df.groupby('department')['salary'].transform('mean')
print(df4)

df5=df.groupby('department').filter(lambda x:len(x)>2)
print(df5)

df.pivot_table(index='department',columns='experience',values='salary',aggfunc='mean')

df6=df['dept_avg_salary']=df.groupby('department')['salary'].transform('mean')
print(df6)

df.groupby('department').agg({
    'salary':'sum',
    'experience':'mean'

  })

df7=df.groupby('department')
```
