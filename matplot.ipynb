```python
import matplotlib.pyplot as plt
import numpy as np
data=[
    30,35,40,42,45,47,48,49,
    50,50,51,52,53,55,57,60,
    62,65,70,75
]
plt.figure(figsize=(5,7))
plt.hist(
    data,bins=5,color="green",edgecolor="black",alpha=0.5,density=True,
    range=(30,80)
)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Histogram")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x=np.array([0,1,2,3])
y=np.array([10,20,30,40])
plt.subplot(2,1,1)
plt.plot(x,y,color='yellow')

x=np.array([0,1,2,3])
y=np.array([3,8,1,10])
plt.subplot(2,1,2)
plt.plot(x,y,c="grey")

x=np.array([0,1,2,3])
y=np.array([10,20,30,40])
plt.subplot(2,3,3)
plt.plot(x,y,linestyle="-.",color='green')

x=np.array([0,1,2,3])
y=np.array([3,8,1,10])
plt.subplot(2,3,4)
plt.scatter(x,y,c="magenta",alpha=0.9)
plt.show()
```
