```python
def val_crop_yield():
  crops=["wheat","rice","maize"]
  yield_per_acre=[20,30,25]
  area=10


  for i in range(len(crops)):
   total_yield=yield_per_acre[i]*area
   print(f"{crops[i]} total yield = {total_yield} quintals")

val_crop_yield()

```

```python
def task2_rainfall_analysis():
   rainfall=(120,25,45,67,47,96)

   highest = max(rainfall)
   lowest =min(rainfall)
   total = sum(rainfall)
   average = total/len(rainfall)


   print(f"The highest rainfall = {highest}mm")
   print(f"The lowest rainfall = {lowest}mm")
   print(f"The total rainfall = {total}mm")
   print(f"The average rainfall = {average}mm")

task2_rainfall_analysis()
```

```python
def profit_loss_calc():

  yield_value=float(input("enter yield:"))
  price=float(input("enter price per unit:"))
  cost=float(input("enter  total cost:"))

  profit=(yield_value*price)-cost

  if profit>0:
    print(f"profit={profit}")
  elif profit<0:
    print(f"loss={abs(profit)}")
  else:
    print("break-even")

profit_loss_calc()
```
