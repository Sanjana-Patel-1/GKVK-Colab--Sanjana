```python
import numpy as np

# Assuming 'array' is a NumPy array or a list
# And 'total_sum' is intended to accumulate values

total_sum = 0 # Initialize sum
my_array = np.arange(100) # Example: create a NumPy array of 100 elements

for i in range(100):
    total_sum += my_array[i]

print(total_sum)
```
