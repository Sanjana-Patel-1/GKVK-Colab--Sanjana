```python
print("\n===crop quality grading===")

weight=85#grams per fruit
blemishes=0
colour_score=9 #out of 10

if weight>=80 and blemishes==0 and colour_score>=8:
  grade="A(PREMIUM)"
elif weight>=60 and blemishes<=2 and colour_score>=6:
  grade="B(STANDARD)"
elif weight>=40:
  grade="C(PROCESSING)"
else:
  grade="(reject)"

print(f"crop grade:{grade}")
```
